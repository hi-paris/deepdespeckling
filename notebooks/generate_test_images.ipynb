```python
import os
import numpy as np 
import cv2
```

```python
def generate_sar_image(save_path:str, H=605, W=1600, save_image=True):
    """
    Generates a random SAR image with two channels (e.g., amplitude and phase) and saves it as a .npy file.
    
    Args:
        save_path (str): Path to save the generated SAR image.
        H (int): Height of the SAR image.
        W (int): Width of the SAR image.
        if_save (bool): Whether to save the generated image as a .npy file.

    Returns:
        sar_image (np.ndarray): Generated SAR image with shape (H, W, 2).
    """
    sar_image = np.random.rand(H, W, 2).astype(np.float32)
    if save_path and save_image:
        if save_path.endswith(".npy"):
            np.save(save_path, sar_image)
            print(f"SAR image saved at {save_path}")
        else:
            raise ValueError("Save path must end with .npy extension.")
    else:
        raise ValueError("Save path is not provided or if_save is False.")
        
    return sar_image


def generate_amplitude_image(image_path: str, save_path: str, save_image=True):
    """    Extracts the amplitude channel from a SAR image and saves it as a .npy file. 
    Args:
        image_path (str): Path to the input SAR image file.
        save_path (str): Path to save the extracted amplitude image.
    """
    image = cv2.imread(image_path) 
    amplitude_image = np.array(image[:, :, 0].astype(np.float32)) 

    if save_image and save_path:
        np.save(save_path, amplitude_image)
    else:
        raise ValueError("Save path is not provided or save_image is False.")
    
    return amplitude_image
   

def generate_merlin_image(save_path: str, H=256, W=256, save_image=True):
    """ Generate image for Merlin method and saves it as a .npy file. 
    
    Args:
        image_path (str): Path to the input SAR image file.
        save_path (str): Path to save the extracted Merlin image.

    Returns:
        merlin_image (np.ndarray): Generated complex-valued SAR image with shape (H, W, 2).
    """
    # Create dummy complex-valued SAR image: shape (128, 128, 2)
    H, W = 256, 256
    real_part = np.random.rand(H, W).astype(np.float32)
    imag_part = np.random.rand(H, W).astype(np.float32)

    # Stack as last dim: (H, W, 2)
    merlin_image =np.stack((real_part, imag_part), axis=-1)
    if save_image and save_path:
        np.save(save_path, merlin_image)
    else:
        raise ValueError("Save path is not provided or save_image is False.")
    
    return merlin_image


def make_dir(dir_path: str):
    """Creates a directory if it does not exist."""
    if not os.path.exists(dir_path):
        os.makedirs(dir_path, exist_ok=True)
        print(f"Directory created: {dir_path}")
    else:
        print(f"Directory already exists: {dir_path}")
        
```

```python
# Defininng paths
current_path = os.getcwd()
data_dir = os.path.join(os.path.dirname(current_path), "data")                                  
data_sar_dir = os.path.join(data_dir, "data_sar")
data_amplitude_dir = os.path.join(data_dir, "data_amplitude")
data_merlin_dir = os.path.join(data_dir, "data_merlin")
make_dir(data_sar_dir)
make_dir(data_amplitude_dir)
make_dir(data_merlin_dir)
```
